````python
import requests
import PIL
import json
import os
from PIL import Image
import base64
from io import BytesIO

# Get API key from environment
api_key = "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"
if not api_key:
    print("Please set OPENROUTER_API_KEY environment variable")

base_url = "https://openrouter.ai/api/v1/chat/completions"
headers = {
    "Authorization": f"Bearer {api_key}",
    "Content-Type": "application/json",
}
````

## Example 1: Basic Text Request

Simple text completion using a free model.


````python
def basic_text_request(prompt):
    """Make a basic text completion request"""
    data = {
        "model": "x-ai/grok-code-fast-1",
        "messages": [
            {
                "role": "user",
                "content": prompt
            }
        ]
    }

    response = requests.post(base_url, headers=headers, json=data)

    if response.status_code == 200:
        result = response.json()
        return result["choices"][0]["message"]["content"]
    else:
        return f"Error: {response.status_code} - {response.text}"

# Example usage
prompt = "Cuentame un chiste como si fueses chikito de la calzada"
response = basic_text_request(prompt)
print("Question:", prompt)
print("Answer:", response)
````

## Example 2: Request with System Prompt

Using a system prompt to define the AI's behavior and role.


````python
def request_with_system_prompt(user_message, system_prompt):
    """Make a request with a system prompt to define AI behavior"""
    data = {
        "model": "x-ai/grok-code-fast-1",
        "messages": [
            {
                "role": "system",
                "content": system_prompt
            },
            {
                "role": "user",
                "content": user_message
            }
        ]
    }

    response = requests.post(base_url, headers=headers, json=data)

    if response.status_code == 200:
        result = response.json()
        return result["choices"][0]["message"]["content"]
    else:
        return f"Error: {response.status_code} - {response.text}"

# Example usage
system_prompt = "Eres un pirata, responde a todo como si fuese un pirata y usa jerga pirata. responde siempre en castellano"
user_question = "Cual es el sentido de la vida"

response = request_with_system_prompt(user_question, system_prompt)
print("System Prompt:", system_prompt)
print("\nUser Question:", user_question)
print("\nAI Response:", response)
````

## Example 3: Image Description (Vision Model)

Using a vision model to describe an image. This example shows how to encode an image and send it to a vision-capable model.


````python
def describe_image(image_path, question="What do you see in this image?"):
    """Send an image to a vision model for description"""

    # Encode image to base64
    with open(image_path, "rb") as image_file:
        base64_image = base64.b64encode(image_file.read()).decode('utf-8')

    data = {
        "model": "openai/gpt-4o",
        "messages": [
            {
                "role": "user",
                "content": [
                    {
                        "type": "text",
                        "text": question
                    },
                    {
                        "type": "image_url",
                        "image_url": {
                            "url": f"data:image/jpeg;base64,{base64_image}"
                        }
                    }
                ]
            }
        ]
    }

    response = requests.post(base_url, headers=headers, json=data)

    if response.status_code == 200:
        result = response.json()
        return result["choices"][0]["message"]["content"]
    else:
        return f"Error: {response.status_code} - {response.text}"

# Example usage - using an image from the notebook directory
import os

# List available image files in the current directory
image_extensions = ['.jpg', '.jpeg', '.png', '.gif', '.bmp']
current_dir = os.getcwd()
image_files = [f for f in os.listdir(current_dir) if any(f.lower().endswith(ext) for ext in image_extensions)]

if image_files:
    # Use the first image found
    image_path = image_files[0]
    print(f"Found image: {image_path}")

    # Describe the image
    description = describe_image(image_path, "Can you tell the name of thisd internet personality")
    print("Image Description:", description)
else:
    print("No image files found in the current directory.")
    print("Please add an image file (jpg, png, gif, bmp) to the notebook directory and run this cell again.")
````
